# Outlier Detection

Outliers were evaluated using the Interquartile Range (IQR) method.

Outliers were not automatically removed because extreme customer behavior may represent legitimate business activity.

The objective is to identify unusual observations and determine whether they are data errors or valid customer behaviors.

```python
# Numerical columns to evaluate

numeric_cols = [
    'age',
    'listening_time',
    'songs_played_per_day',
    'skip_rate',
    'ads_listened_per_week'
]

for col in numeric_cols:

    Q1 = spotify_df[col].quantile(0.25)
    Q3 = spotify_df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = spotify_df[
        (spotify_df[col] < lower_bound) |
        (spotify_df[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} outliers")
```

## Outlier Findings

Potential outliers were evaluated using the Interquartile Range (IQR) method.

No outliers were identified for:

- age
- listening_time
- songs_played_per_day
- skip_rate

The variable ads_listened_per_week contained 1,683 observations identified as potential outliers.

These values were not removed because they appear to represent legitimate customer behavior rather than data entry errors. The distribution is likely influenced by differences between subscription types, where premium subscribers may receive no advertisements while free-tier users may receive substantially more.

Therefore, all observations were retained for future analysis.
